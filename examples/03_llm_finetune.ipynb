# LLM Finetuning Walkthrough
Use this notebook as a lightweight checklist for preparing a custom fine-tuned model with PromptCraft tooling.


## Prerequisites
- Install the optional OpenAI SDK: `pip install openai`
- Configure credentials via `PROMPTCRAFT_OPENAI_API_KEY` (and optional base URL or organization overrides)
- Gather high-quality prompt/completion pairs that reflect the behaviour you expect from the tuned model


```python
from promptcraft.models import Prompt

training_plan = [
    {
        "input": Prompt(text="Summarize release notes for executives."),
        "ideal": "Provide a three-bullet executive summary.",
    },
    {
        "input": Prompt(text="Draft a customer apology email using issue details."),
        "ideal": "Write a concise, empathetic apology and remediation plan.",
    },
]
training_plan
```

## Upload curated data
Use your storage backend or the OpenAI files API to upload JSONL documents that mirror the `training_plan` skeleton.


```python
from promptcraft.ai_client import AIClient

client = AIClient(model="gpt-4o-mini")
job_request = {
    "training_file": "file-abc123",
    "validation_file": "file-def456",
    "model": client.model,
}
job_request
```

## Track experiments
Capture dataset references and prompt variants alongside PromptCraft sessions so that results remain reproducible.


```python
from promptcraft.models import Session

session = Session(
    user_id="finetune-pipeline",
    prompts=[entry["input"].text for entry in training_plan],
)
session
```

## Next steps
1. Submit the request with `openai.fine_tuning.jobs.create(**job_request)` once files are uploaded.
2. Monitor job status using `openai.fine_tuning.jobs.retrieve(job_id)` and log checkpoints in PromptCraft.
3. Evaluate the tuned model with the scoring helpers in `examples/02_scoring_demo.ipynb`.
